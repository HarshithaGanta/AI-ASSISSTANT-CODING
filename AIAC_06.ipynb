**Task 1: Classes (Student Class)**

Generate a Python class named Student with attributes name, roll number, and branch. Add a method to display student details.

```python
class Student:
    def __init__(self, name, roll_number, branch):
        self.name = name
        self.roll_number = roll_number
        self.branch = branch

    def display_details(self):
        print(f"Student Name: {self.name}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Branch: {self.branch}")
```

```python
# Example usage:
student1 = Student("Alice", "A101", "Computer Science")
student1.display_details()
```

**Task 2: Loops (Multiples of a Number)**

**Prompt 1 (For Loop):**

Generate a Python function to print the first 10 multiples of a given number using a for loop.

```python
def print_multiples(number):
    print(f"First 10 multiples of {number}:")
    for i in range(1, 11):  # Loop 10 times for the first 10 multiples
        print(number * i)
```

```python
# Example usage:
print_multiples(7)
print_multiples(12)
```

**Prompt 2(While loop):**

Generate a Python function to print the first 10 multiples of a given number using a while loop

```python
def print_multiples_while(number):
    print(f"First 10 multiples of {number} (using while loop):")
    count = 1
    while count <= 10:
        print(number * count)
        count += 1
```

```python
# Example usage:
print_multiples_while(8)
print_multiples_while(15)
```

**Analysis**

for loop is concise and best when iteration count is fixed.

while loop provides manual control over looping.

Both implementations are correct.

**Task 3: Conditional Statements – Age Classification**


**Prompt 1 (if-elif-else)**

Generate a Python function using nested if-elif-else statements to classify age as child, teenager, adult, or senior.

```python
def classify_age(age):
    if age < 0:
        return "Invalid Age"
    elif age <= 12:
        return "Child"
    elif age <= 17:
        return "Teenager"
    elif age <= 64:
        return "Adult"
    else:
        return "Senior"
```

```python
# Example usage:
print(f"Age 5: {classify_age(5)}")
print(f"Age 15: {classify_age(15)}")
print(f"Age 30: {classify_age(30)}")
print(f"Age 70: {classify_age(70)}")
print(f"Age -2: {classify_age(-2)}")
```

**Prompt 2 (Simplified Logic)**

Generate an alternative age classification using simplified conditional logic.

```python
def classify_age_simplified(age):
    if age < 0:
        return "Invalid Age"
    elif age < 13:
        return "Child" # Ages 0-12
    elif age < 18:
        return "Teenager" # Ages 13-17
    elif age < 65:
        return "Adult" # Ages 18-64
    else:
        return "Senior" # Ages 65+
```

```python
# Example usage:
print(f"Age 5: {classify_age_simplified(5)}")
print(f"Age 15: {classify_age_simplified(15)}")
print(f"Age 30: {classify_age_simplified(30)}")
print(f"Age 70: {classify_age_simplified(70)}")
print(f"Age -5: {classify_age_simplified(-5)}")
```

**Analysis**

Output remains correct.

**Task 4: For and While Loops – Sum of First n Numbers**

**Prompt 1 (For Loop)**

Generate a Python function to calculate the sum of the first n natural numbers using a for loop

```python
def sum_first_n_for(n):
    if n < 1:
        return "Please enter a positive integer for n."
    total_sum = 0
    for i in range(1, n + 1):
        total_sum += i
    return total_sum
```

```python
# Example usage:
print(f"Sum of first 5 natural numbers (for loop): {sum_first_n_for(5)}")
print(f"Sum of first 10 natural numbers (for loop): {sum_first_n_for(10)}")
print(f"Sum of first 0 natural numbers (for loop): {sum_first_n_for(0)}")
```

**Prompt 2 (While Loop)**

Generate a Python function to calculate the sum of the first n natural numbers using a while loop.

```python
def sum_first_n_while(n):
    if n < 1:
        return "Please enter a positive integer for n."
    total_sum = 0
    count = 1
    while count <= n:
        total_sum += count
        count += 1
    return total_sum
```

```python
# Example usage:
print(f"Sum of first 5 natural numbers (while loop): {sum_first_n_while(5)}")
print(f"Sum of first 10 natural numbers (while loop): {sum_first_n_while(10)}")
print(f"Sum of first 0 natural numbers (while loop): {sum_first_n_while(0)}")
```

**Prompt 3 (Mathematical Formula)**

Generate a Python function to calculate the sum of the first n natural numbers using a mathematical formula.

```python
def sum_first_n_formula(n):
    if n < 1:
        return "Please enter a positive integer for n."
    # Formula for the sum of the first n natural numbers: n * (n + 1) / 2
    return int(n * (n + 1) / 2)
```

```python
# Example usage:
print(f"Sum of first 5 natural numbers (formula): {sum_first_n_formula(5)}")
print(f"Sum of first 10 natural numbers (formula): {sum_first_n_formula(10)}")
print(f"Sum of first 0 natural numbers (formula): {sum_first_n_formula(0)}")
```

**Analysis**

Loop methods are easy to understand.

Formula method is most efficient.

All approaches produce correct results.

**Task 5: Classes – Bank Account Class**

Generate a Python class named `BankAccount` with attributes `account_holder_name` and `balance`. Include methods to `deposit` money, `withdraw` money, and `display_balance`.

```python
class BankAccount:
    def __init__(self, account_holder_name, initial_balance=0):
        self.account_holder_name = account_holder_name
        self.balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self.balance >= amount:
                self.balance -= amount
                print(f"Withdrew {amount}. New balance: {self.balance}")
            else:
                print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def display_balance(self):
        print(f"Account Holder: {self.account_holder_name}, Current Balance: {self.balance}")
```

```python
# Example usage:
my_account = BankAccount("John Doe", 1000)
my_account.display_balance()
my_account.deposit(500)
my_account.withdraw(200)
my_account.withdraw(1500) # Attempt to withdraw more than balance
my_account.deposit(-100) # Attempt to deposit negative amount
my_account.display_balance()
```

**Explanation**


Constructor initializes account details.

Deposit increases balance.

Withdraw checks balance before deduction.

Code is clear, safe, and functional.
