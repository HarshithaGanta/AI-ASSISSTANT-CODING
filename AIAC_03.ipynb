**Question 1: Zero-Shot Prompting – Palindrome Number**

**Zero-Shot Prompt**:

Generate a Python function to check whether a given number is a palindrome.

```python
def is_palindrome_number(number):
    # Convert the number to a string to easily reverse it
    num_str = str(number)
    # Compare the string with its reversed version
    return num_str == num_str[::-1]

# Example usage:
print(f"Is 121 a palindrome? {is_palindrome_number(121)}")
print(f"Is 123 a palindrome? {is_palindrome_number(123)}")
print(f"Is 1221 a palindrome? {is_palindrome_number(1221)}")
print(f"Is -121 a palindrome? {is_palindrome_number(-121)}")
print(f"Is 0 a palindrome? {is_palindrome_number(0)}")
```

**Analysis**

Logic is correct for positive integers.

Does not explicitly handle negative numbers.

Simple and readable zero-shot solution.

**Question 2: One-Shot Prompting – Factorial Calculation**

**One-Shot Prompt**

Generate a Python function to compute factorial.
Example: Input: 5 → Output: 120

```python
def calculate_factorial_one_shot(n):
    if not isinstance(n, int):
        return "Input must be an integer."
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0:
        return 1
    else:
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result

# Example usage based on the prompt:
print(f"Factorial of 5: {calculate_factorial_one_shot(5)}")

# Additional examples to test the function:
print(f"Factorial of 0: {calculate_factorial_one_shot(0)}")
print(f"Factorial of 7: {calculate_factorial_one_shot(7)}")
print(f"Factorial of -3: {calculate_factorial_one_shot(-3)}")
print(f"Factorial of 4.5: {calculate_factorial_one_shot(4.5)}")
```

**Comparison with Zero-Shot**

One-shot prompting improves clarity.

Correct handling of loop boundaries.

More structured than zero-shot output.

**Question 3: Few-Shot Prompting – Armstrong Number**

**Few-Shot Prompt**:

Check whether a number is an Armstrong number.
Input: 153 → Output: Armstrong Number
Input: 370 → Output: Armstrong Number
Input: 123 → Output: Not an Armstrong Number

```python
def is_armstrong_number(number):
    if not isinstance(number, int) or number < 0:
        return "Input must be a non-negative integer."
    if number == 0:
        return True # 0 is generally considered an Armstrong number

    # Convert number to string to find its length (number of digits)
    num_str = str(number)
    num_digits = len(num_str)
    sum_of_powers = 0
    for digit_char in num_str:
        digit = int(digit_char)
        sum_of_powers += digit ** num_digits

    # Compare the sum with the original number
    return sum_of_powers == number

print(f"Is 153 an Armstrong number? {is_armstrong_number(153)}")
print(f"Is 370 an Armstrong number? {is_armstrong_number(370)}")
print(f"Is 123 an Armstrong number? {is_armstrong_number(123)}")

# Additional examples to test the function:
print(f"Is 371 an Armstrong number? {is_armstrong_number(371)}") # 3^3 + 7^3 + 1^3 = 27 + 343 + 1 = 371
print(f"Is 9474 an Armstrong number? {is_armstrong_number(9474)}") # 9^4 + 4^4 + 7^4 + 4^4 = 6561 + 256 + 2401 + 256 = 9474
print(f"Is 10 an Armstrong number? {is_armstrong_number(10)}")
print(f"Is -153 an Armstrong number? {is_armstrong_number(-153)}")
print(f"Is 0 an Armstrong number? {is_armstrong_number(0)}")
```

**Analysis**

Few-shot examples guide correct formula usage.

Code structure is more accurate.

Works correctly for boundary values like 0 and 1.

**Question 4: Context-Managed Prompting – Number Classification**

**Context-Managed Prompt**

Write an optimized Python program to classify a number as prime, composite, or neither.
Validate input and optimize for efficiency.

```python
import math

def classify_number(number):
    if not isinstance(number, int):
        return "Invalid Input"

    if number <= 1:
        return "Neither"

    if number == 2:
        return "Prime"

    if number % 2 == 0:
        return "Composite"

    # Check for odd divisors from 3 up to the square root of the number
    for i in range(3, int(math.sqrt(number)) + 1, 2):
        if number % i == 0:
            return "Composite"

    return "Prime"

# Test cases based on the prompt's constraints and typical scenarios:
print(f"Classify 2: {classify_number(2)}")         # Expected: Prime
print(f"Classify 3: {classify_number(3)}")         # Expected: Prime
print(f"Classify 4: {classify_number(4)}")         # Expected: Composite
print(f"Classify 0: {classify_number(0)}")         # Expected: Neither
print(f"Classify 1: {classify_number(1)}")         # Expected: Neither
print(f"Classify 97: {classify_number(97)}")       # Expected: Prime
print(f"Classify 100: {classify_number(100)}")     # Expected: Composite
print(f"Classify 121: {classify_number(121)}")     # Expected: Composite (11*11)
print(f"Classify 127: {classify_number(127)}")     # Expected: Prime
print(f"Classify 5.5: {classify_number(5.5)}")     # Expected: Invalid Input
print(f"Classify 'abc': {classify_number('abc')}") # Expected: Invalid Input
print(f"Classify -7: {classify_number(-7)}")       # Expected: Neither
```

**Analysis**

Input validation handled correctly.

Efficient logic using √n optimization.

Better than earlier prompting strategies.

**Question 5: Zero-Shot Prompting – Perfect Number**

**Zero-Shot Prompt**

Generate a Python function to check whether a number is a perfect number.

```python
def is_perfect_number(number):
    if not isinstance(number, int) or number <= 0:
        return False
    if number == 1:
        return False

    sum_divisors = 1  # Start with 1, as 1 is a proper divisor for all numbers > 1
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            sum_divisors += i
            if i * i != number:
                sum_divisors += number // i

    return sum_divisors == number

print(f"Is 6 a perfect number? {is_perfect_number(6)}") # Expected: True
print(f"Is 28 a perfect number? {is_perfect_number(28)}") # Expected: True
print(f"Is 496 a perfect number? {is_perfect_number(496)}") # Expected: True
print(f"Is 10 a perfect number? {is_perfect_number(10)}") # Expected: False
print(f"Is 1 a perfect number? {is_perfect_number(1)}") # Expected: False
print(f"Is 0 a perfect number? {is_perfect_number(0)}") # Expected: False
print(f"Is -28 a perfect number? {is_perfect_number(-28)}") # Expected: False
print(f"Is 7 a perfect number? {is_perfect_number(7)}") # Expected: False (prime number)
```

**Analysis**


Correct logic for perfect numbers.

Inefficient for large numbers.

No optimization or early termination.

**Question 6: Few-Shot Prompting – Even or Odd with Validation**

**Few-Shot Prompt**

Determine whether a number is even or odd.

Input: 8 → Even

Input: 15 → Odd

Input: 0 → Even

Include proper input validation.

```python
def classify_even_odd(number):
    if not isinstance(number, int):
        return "Invalid Input"
    elif number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(f"Classify 8: {classify_even_odd(8)}")
print(f"Classify 15: {classify_even_odd(15)}")
print(f"Classify 0: {classify_even_odd(0)}")

# Additional examples to test validation and other cases:
print(f"Classify 7: {classify_even_odd(7)}")
print(f"Classify -4: {classify_even_odd(-4)}")
print(f"Classify -9: {classify_even_odd(-9)}")
print(f"Classify 3.5: {classify_even_odd(3.5)}")
print(f"Classify 'hello': {classify_even_odd('hello')}")
```

**Analysis**

Few-shot prompting improves validation handling.

Handles negative numbers correctly.

Clear and user-friendly output.

Zero-shot prompts generate basic solutions.

One-shot and few-shot prompts significantly improve correctness.

Context-managed prompting produces optimized and robust code.

Prompt engineering directly impacts code quality and reliability.
