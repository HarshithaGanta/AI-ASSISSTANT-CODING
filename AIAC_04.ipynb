## **Lab 4: Advanced Prompt Engineering – Zero-Shot, One-Shot, and Few-Shot Techniques**

### **Task 1: Zero-Shot Prompting – Leap Year Check**

Generate a Python function that checks whether a given year is a leap year.

```python
def is_leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return True
    return False

year = int(input())
print(is_leap_year(year))
```

### **Task 2: One-Shot Prompting – Centimeters to Inches**

Convert centimeters to inches.

Example: Input: 10 cm → Output: 3.94 inches

```python
def cm_to_inches(cm):
    return cm * 0.394

cm = float(input())
print(cm_to_inches(cm))
```

### **Task 3: Few-Shot Prompting – Name Formatting**

Format full names as “Last, First”.

Examples:

"John Smith" → "Smith, John"

"Anita Rao" → "Rao, Anita"

```python
def format_name(name):
    parts = name.split()
    return parts[-1] + ", " + parts[0]

name = input()
print(format_name(name))
```

### **Task 4: Zero-Shot vs Few-Shot – Vowel Count**

**Zero-Shot Code**

```python
def count_vowels(s):
    count = 0
    for ch in s:
        if ch in "aeiouAEIOU":
            count += 1
    return count

print(count_vowels(input()))
```

**Few-Shot Code**

```python
def count_vowels(s):
    return sum(1 for ch in s if ch.lower() in "aeiou")

print(count_vowels(input()))
```

### **Task 5: Few-Shot Prompting – File Handling**

Read a .txt file and count number of lines.
Example:
File with 3 lines → Output: 3

```python
def count_lines(filename):
    with open(filename, "r") as f:
        return len(f.readlines())

print(count_lines("sample.txt"))
```
