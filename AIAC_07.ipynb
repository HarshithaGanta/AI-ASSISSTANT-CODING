### **Task 1: Syntax Error – Missing Parentheses**

Identify the syntax error in the following Python code and correct it. Explain the error and verify the fix using assert test cases.

**Buggy Code**

```python

def greet():
    print "Hello, AI Debugging Lab!"

greet()
```

**Observed Error**

SyntaxError: Missing parentheses in call to 'print'


**Corrected Code**

```python
def greet():
    return "Hello, AI Debugging Lab!"

assert greet() == "Hello, AI Debugging Lab!"
assert isinstance(greet(), str)
assert greet().startswith("Hello")

print(greet())
```

### **Task 2: Logic Error – Using = Instead of == in if Condition**

Explain why using = instead of == causes an error in the if-condition and fix the code. Validate using assert statements.

**Buggy Code**

```python
def check_number(n):
    if n = 10:
        return "Ten"
    else:
        return "Not Ten"
```

**Observed Error**

```python
SyntaxError: cannot assign to expression
```

**Corrected Code**

```python
def check_number(n):
    if n == 10:
        return "Ten"
    else:
        return "Not Ten"

assert check_number(10) == "Ten"
assert check_number(5) == "Not Ten"
assert check_number(0) == "Not Ten"

print(check_number(10))
```

### **Task 3: Runtime Error – File Not Found**

Modify the following code to handle file-not-found errors safely using try-except and provide user-friendly messages.

**Buggy Code**

```python
def read_file(filename):
    with open(filename, 'r') as f:
        return f.read()

print(read_file("nonexistent.txt"))
```

**Observed Error**

FileNotFoundError: [Errno 2] No such file or directory


**Corrected Code**

```python
def read_file(filename):
    if not filename:
        return "Error: Invalid file path"
    try:
        with open(filename,'r') as f:
            return f.read()
    except FileNotFoundError:
        return "Error: File not found"

assert read_file("nonexistent.txt")=="Error: File not found"
assert read_file("")=="Error: Invalid file path"
assert isinstance(read_file("nonexistent.txt"),str)

print(read_file("nonexistent.txt"))
```

### **Task 4: Calling a Non-Existent Method**

Analyze the following class and fix the error caused by calling a non-existent method. Add test cases.

**Buggy Code**

```python
class Car:
    def start(self):
        return "Car started"

my_car = Car()
print(my_car.drive())
```

**Observed Error**

AttributeError: 'Car' object has no attribute 'drive'


**Corrected Code (Method Added)**

```python
class Car:
    def start(self):
        return "Car started"

    def drive(self):
        return "Car is driving"

my_car = Car()

assert my_car.start() == "Car started"
assert my_car.drive() == "Car is driving"
assert isinstance(my_car.drive(), str)

print(my_car.drive())
```

### **Task 5: TypeError – Mixing String and Integer**

The following code raises a TypeError. Provide two solutions: one using type casting and one using string concatenation. Validate both.

**Buggy Code**

```python
def add_five(value):
    return value + 5

print(add_five("10"))
```

**Observed Error**

TypeError: can only concatenate str (not "int") to str


**Solution 1: Type Casting**

```python
def add_five(value):
    return int(value) + 5

assert add_five(10) == 15
assert add_five("10") == 15
assert add_five(0) == 5

print(add_five("10"))
```

**Solution 2: String Concatenation**

```python
def add_five(value):
    return str(value) + "5"

assert add_five("10") == "105"
assert add_five(3) == "35"
assert isinstance(add_five(1), str)

print(add_five("10"))
```
