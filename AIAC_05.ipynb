### **Task 2: Privacy & Security in File Handling**

```python
import hashlib

name = input()
email = input()
password = input()

hashed_password = hashlib.sha256(password.encode()).hexdigest()

with open("users.txt", "a") as f:
    f.write(name + "," + email + "," + hashed_password + "\n")

print("User data stored securely")

```

### **Task 3: Transparency in Algorithm Design**

```python
def armstrong(n):
    temp = n
    s = 0
    d = len(str(n))

    while temp > 0:
        r = temp % 10
        s += r ** d
        temp //= 10

    return s == n

num = int(input())
print(armstrong(num))
```

### **Task 4: Sorting Algorithms Comparison**

**Bubble Sort Code**

```python
def bubble_sort(a):
    for i in range(len(a)):
        for j in range(len(a)-1-i):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a

print(bubble_sort([5, 3, 1, 4, 2]))
```

**Quick Sort Code**

```python
def quick_sort(a):
    if len(a) <= 1:
        return a
    p = a[0]
    l = [x for x in a[1:] if x <= p]
    r = [x for x in a[1:] if x > p]
    return quick_sort(l) + [p] + quick_sort(r)

print(quick_sort([5, 3, 1, 4, 2]))
```
