### Factorial Calculation using a Loop

This code block calculates the factorial of a given number using a `for` loop. It takes an integer input, initializes a `factorial` variable to 1, and then multiplies it by each number from 1 up to the input number. It also handles edge cases for negative numbers and zero.

```python
# Input a number
num = int(input("Enter a non-negative integer: "))

# Check if the number is negative
if num < 0:
    print("Factorial is not defined for negative numbers.")
# Check if the number is 0
elif num == 0:
    print("The factorial of 0 is 1.")
# Calculate factorial for positive numbers
else:
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f"The factorial of {num} is {factorial}.")
```

### Factorial Function with Main Execution

This code block defines a Python function `calculate_factorial` that takes a non-negative integer as input and returns its factorial. It includes error handling for negative numbers and a base case for zero. The main part of the script then prompts the user for input and calls this function, displaying the result.

```python
def calculate_factorial(num):
    """
    Calculates the factorial of a non-negative integer.

    Args:
        num: A non-negative integer.

    Returns:
        The factorial of the number if num >= 0, otherwise prints an error message.
    """
    if num < 0:
        print("Factorial is not defined for negative numbers.")
        return None
    elif num == 0:
        return 1
    else:
        factorial = 1
        for i in range(1, num + 1):
            factorial *= i
        return factorial

# Main part of the program
if __name__ == "__main__":
    try:
        input_num = int(input("Enter a non-negative integer: "))
        result = calculate_factorial(input_num)
        if result is not None:
            print(f"The factorial of {input_num} is {result}.")
    except ValueError:
        print("Invalid input. Please enter an integer.")
```

**Short Note**

Using functions improves reusability and clarity.
The same logic can be reused in multiple programs without rewriting code.

**Task 5: Iterative vs Recursive Thinking**

Iterative Version (AI-Generated)

```python
def factorial_iterative(n):
    result=1
    for i in range(1,n+1):
        result*=i
    return result

num=int(input("Enter a number: "))
print("Factorial is:",factorial_iterative(num))
```

Recursive Version (AI-Generated)

```python
def factorial_recursive(n):
    if n==0 or n==1:
        return 1
    return n*factorial_recursive(n-1)

num=int(input("Enter a number: "))
print("Factorial is:",factorial_recursive(num))

```

**Execution Flow Explanation**

The iterative version uses a loop to multiply values step-by-step.
The recursive version repeatedly calls itself until the base condition is reached.


**Comparison**

Readability: Iterative is easier for beginners

Stack Usage: Recursive uses more memory

Performance: Iterative is faster

Recursion Not Recommended: For very large inputs due to stack overflow
